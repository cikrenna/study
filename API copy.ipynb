```python
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Configurazione dell'API
API_KEY = "727b724ba0f32deefd6fca99f0f63434"          # <-- inserisci la tua chiave
HEADERS = {"api-key": API_KEY}
BASE_URL = "http://www.tng-project.org/api/"

def get(url, params=None):
    r = requests.get(url, params=params, headers=HEADERS)
    r.raise_for_status()
    if r.headers['content-type'] == 'application/json':
        return r.json()
    return r

# Test: otteniamo la lista delle simulazioni disponibili
simulations = get(BASE_URL)
```

```python
# Stampiamo i nomi delle simulazioni disponibili
print("Simulazioni disponibili:")
for sim in simulations['simulations']:
    print(" -", sim['name'])
```
